# Actor mandate and analytical question

**Actor 4: Global Youth & Future Generations Coalition**

This notebook is the final model-based synthesis for the Global Youth & Future Generations Coalition. The actor mandate treats **1.5 degC as the political red line** for climate safety. The saved optimisation and robustness workflow still uses the course/JUSTICE 2 degC threshold where that is built into the Assignment 5-8 results; the 1.5 degC evidence is used as an Actor 4 diagnostic benchmark, not as a claim that the original optimisation was run at 1.5 degC.

The central analytical question is:

> How does welfare framing change which damages, trade-offs, and policies become visible to the optimiser, and what does that imply for a coalition representing young people and future generations?

A welfare function does **not** change climate physics directly. For a fixed policy and physical scenario, temperature follows the JUSTICE climate-economy dynamics. The welfare framing changes how future damages, regional losses, and abatement burdens are counted, which changes optimisation pressure and the policies that appear attractive.

The notebook follows the assignment logic: uncertainty -> sensitivity -> scenario discovery -> problem formulation -> many-objective optimisation -> convergence -> Pareto trade-offs -> actor-specific selection -> robustness -> political advice.


## Welfare framing and visibility

The final recommendation compares two welfare framings that are politically meaningful for Actor 4:

| Framing | Role in this notebook | Actor interpretation |
|---|---|---|
| **Prioritarian / near-zero-discount framing** | Actor-aligned main lens | Gives stronger visibility to future and vulnerable populations, so delayed damages count more heavily in policy search and regret calculations. |
| **Utilitarian / standard economic-discount framing** | Rival comparator | Gives more weight to aggregate present welfare and near-term costs, making delayed mitigation or lower immediate abatement burden appear more acceptable. |

This contrast is used to ask whose losses become visible under each framing. It is not treated as a direct physical climate switch; physical outcomes change only through the policies selected under each framing and through the scenario/ensemble assumptions used to evaluate them.

### Welfare comparison used to anchor the mandate

Before moving through the assignment workflow, the notebook first makes the welfare-framing contrast concrete. The cells below compare how the same illustrative climate-policy context is read through Prioritarian and Utilitarian welfare accounting. This motivates why the Youth Coalition cannot treat the welfare function as a neutral technical choice.


```python
# PRIORITARIAN vs UTILITARIAN single-run comparison using official JUSTICE welfare defaults.
# Scenario-design choice: SSP2, median ECS, deterministic run, moderate abatement (ecr_plateau=0.5).

DEFAULT_COMPARISON_SETUP = {
    "ssp": 2,
    "ecs_pct": 50.0,
    "ecr_plateau": 0.5,
}

def run_justice_welfare(welfare_fn, ssp=2, ecs_pct=50.0, ecr_plateau=0.5):
    """Run JUSTICE and return key outcomes without overriding welfare or damage defaults."""
    with justice_working_directory():
        JUSTICE.hard_reset()
        ensemble_idx = ecs_percentile_to_ensemble(ecs_pct)

        model = JUSTICE(
            start_year=2015,
            end_year=2300,
            timestep=1,
            scenario=ssp,
            climate_ensembles=ensemble_idx,
            stochastic_run=False,
            social_welfare_function=welfare_fn,
        )

        wf = model.welfare_function
        years = np.asarray(model.time_horizon.model_time_horizon)
        tau = np.clip((years - 2015) / (2100 - 2015), 0, 1)
        s_curve = 3 * tau**2 - 2 * tau**3
        ecr = np.clip(np.outer(np.full(len(model.region_list), ecr_plateau), s_curve), 0, 1)

        model.run(emission_control_rate=ecr, endogenous_savings_rate=True)
        datasets = model.evaluate()

        gt = datasets['global_temperature']
        yat_15 = float(years_above_temperature_threshold(gt, 1.5))
        yat_2 = float(years_above_temperature_threshold(gt, 2.0))
        peak = float(np.max(gt))

        # Pattern from the assignment notebooks: only the fourth return value is scalar welfare.
        _, _, _, wl_dam = wf.calculate_welfare(
            datasets['damage_cost_per_capita'], welfare_loss=True
        )
        _, _, _, wl_abt = wf.calculate_welfare(
            datasets['abatement_cost_per_capita'], welfare_loss=True
        )

    return {
        'welfare_loss_damage': float(np.abs(wl_dam)),
        'welfare_loss_abatement': float(np.abs(wl_abt)),
        'yat_15c': yat_15,
        'yat_2c': yat_2,
        'peak_temp': peak,
        'rho': float(wf.pure_rate_of_social_time_preference),
        'eta': float(wf.elasticity_of_marginal_utility_of_consumption),
        'inequality_aversion': float(wf.inequality_aversion),
        'sufficiency_threshold': float(wf.sufficiency_threshold),
        'egality_strictness': float(wf.egality_strictness),
        'welfare_fn': welfare_fn.name,
        'scenario': ssp,
        'ecs_percentile': ecs_pct,
        'ecr_plateau': ecr_plateau,
    }


print('Running PRIORITARIAN with JUSTICE default welfare parameters ...')
res_prior = run_justice_welfare(WelfareFunction.PRIORITARIAN, **DEFAULT_COMPARISON_SETUP)
print(f"  PRIORITARIAN default rho = {res_prior['rho']:.3f}")
print('Running UTILITARIAN with JUSTICE default welfare parameters ...')
res_util = run_justice_welfare(WelfareFunction.UTILITARIAN, **DEFAULT_COMPARISON_SETUP)
print(f"  UTILITARIAN default rho = {res_util['rho']:.3f}")

comparison = pd.DataFrame([res_prior, res_util]).set_index('welfare_fn')
print('\n--- Comparison: PRIORITARIAN vs UTILITARIAN using JUSTICE defaults ---')
display(comparison[[
    'rho', 'eta', 'inequality_aversion', 'sufficiency_threshold', 'egality_strictness',
    'welfare_loss_damage', 'welfare_loss_abatement', 'yat_15c', 'yat_2c', 'peak_temp'
]].round(3))
```

```python
# Bar chart: welfare losses under PRIORITARIAN vs UTILITARIAN defaults.
# Key debate point: under PRIORITARIAN, future generations' damage is not discounted away.

metrics = ['welfare_loss_damage', 'welfare_loss_abatement']
labels_m = ['Welfare loss\n(climate damage)', 'Welfare loss\n(abatement cost)']

prior_vals = [res_prior[m] for m in metrics]
util_vals = [res_util[m] for m in metrics]

x = np.arange(len(metrics))
width = 0.35
fig, ax = plt.subplots(figsize=(7, 5))
ax.bar(
    x - width / 2,
    prior_vals,
    width,
    label=f"PRIORITARIAN default (rho = {res_prior['rho']:.3f})",
    color='#2196F3',
    alpha=0.85,
)
ax.bar(
    x + width / 2,
    util_vals,
    width,
    label=f"UTILITARIAN default (rho = {res_util['rho']:.3f})",
    color='#FF9800',
    alpha=0.85,
)
ax.set_xticks(x)
ax.set_xticklabels(labels_m, fontsize=11)
ax.set_ylabel('Welfare loss (utils)', fontsize=11)
ax.set_title('Welfare losses: PRIORITARIAN vs UTILITARIAN defaults\nScenario: SSP2, median ECS, moderate abatement', fontsize=11)
ax.legend(fontsize=10)
sns.despine(ax=ax)
plt.tight_layout()

img_path = FINAL_PLOTS_DIR / 'fp_welfare_comparison.png'
plt.savefig(img_path, dpi=130, bbox_inches='tight')
plt.show()

print(f'Saved to {img_path}')
print()
print('Debate interpretation:')
print(f"  PRIORITARIAN default rho: {res_prior['rho']:.3f}")
print(f"  UTILITARIAN default rho: {res_util['rho']:.3f}")
print(f"  PRIORITARIAN damage loss: {res_prior['welfare_loss_damage']:.2f}")
print(f"  UTILITARIAN damage loss: {res_util['welfare_loss_damage']:.2f}")
ratio = res_prior['welfare_loss_damage'] / res_util['welfare_loss_damage'] if res_util['welfare_loss_damage'] > 0 else float('nan')
print(f"  Ratio (PRIOR/UTIL): {ratio:.2f}x")
```

**Interpretation - welfare comparison.**

The Prioritarian and Utilitarian runs use the same physical climate trajectory and the official JUSTICE defaults for each welfare function. Under the Utilitarian default (`rho = 0.015`), damages in the far future receive much less pure time-discounting weight. Under the Actor 4 Prioritarian default (`rho = 0.0`), long-run damages remain politically and ethically visible.

For the report, this section supports the claim that discounting is not a neutral technical detail. It changes how the model represents future generations and therefore changes which policies look attractive.

### Discount-weighted regional damage visibility

This figure focuses on the central rival framing for Actor 4: Prioritarian versus Utilitarian. JUSTICE also includes Egalitarian and Sufficientarian welfare functions, but they are not shown in this specific figure because the plot isolates the pure time-discounting component of damage visibility. In that component, Egalitarian belongs to the same low-discount family as Prioritarian, while Sufficientarian belongs to the same higher-discount family as Utilitarian. Their broader welfare logic still differs through inequality aversion, sufficiency thresholding, and egalitarian strictness, but that is not what this figure isolates.

The physical damage-per-capita trajectory is generated once from a deterministic JUSTICE run: SSP2 / scenario index 2, median ECS ensemble, and a moderate abatement S-curve with `ecr_plateau = 0.5`. The figure focuses on `rsas / Rest of South Asia`, one of the two Actor 4-relevant regions and a strong fit for the Youth Coalition narrative because it highlights future-oriented vulnerability in a climate-exposed region. The code still saves the transparent two-region CSV for reproducibility, but the displayed figure uses one region to keep the message readable.

The dashed grey line is the physical damage reference. The blue area shows the damage that remains visible under Prioritarian discounting; the orange area shows the damage that remains visible under Utilitarian discounting. The plotted curves use the official discount vectors returned by the Prioritarian and Utilitarian JUSTICE welfare-function objects. This is a time-discounting visibility plot, not a full comparison of complete welfare aggregation formulas.


```python
# Generate transparent data for regional damage visibility under Prioritarian vs Utilitarian.
# The plotted value is physical damage per capita times the official discount vector from each welfare object.
# This isolates the time-discounting component; it is not a full regional welfare aggregation.

FINAL_RESULTS_DIR = RESULTS_ROOT / "final_project"
FINAL_RESULTS_DIR.mkdir(parents=True, exist_ok=True)
visibility_csv = FINAL_RESULTS_DIR / "prioritarian_vs_utilitarian_regional_damage_visibility.csv"

main_welfare_functions = [
    ("Prioritarian", WelfareFunction.PRIORITARIAN),
    ("Utilitarian", WelfareFunction.UTILITARIAN),
]

region_info = [
    (REGION_IDX_RJAN, REGION_LABEL_RJAN),
    (REGION_IDX_RSAS, REGION_LABEL_RSAS),
]

with justice_working_directory():
    JUSTICE.hard_reset()
    ensemble_idx = ecs_percentile_to_ensemble(50.0)
    physical_model = JUSTICE(
        start_year=2015,
        end_year=2300,
        timestep=1,
        scenario=2,
        climate_ensembles=ensemble_idx,
        stochastic_run=False,
        social_welfare_function=WelfareFunction.UTILITARIAN,
    )

    years_damage = np.asarray(physical_model.time_horizon.model_time_horizon)
    tau = np.clip((years_damage - 2015) / (2100 - 2015), 0, 1)
    s_curve = 3 * tau**2 - 2 * tau**3
    ecr = np.clip(np.outer(np.full(len(physical_model.region_list), 0.5), s_curve), 0, 1)

    physical_model.run(emission_control_rate=ecr, endogenous_savings_rate=True)
    physical_data = physical_model.evaluate()
    physical_damage = np.asarray(physical_data["damage_cost_per_capita"])

    records = []
    for lens_label, lens_enum in main_welfare_functions:
        JUSTICE.hard_reset()
        lens_model = JUSTICE(
            start_year=2015,
            end_year=2300,
            timestep=1,
            scenario=2,
            climate_ensembles=ensemble_idx,
            stochastic_run=False,
            social_welfare_function=lens_enum,
        )
        wf = lens_model.welfare_function
        discount = np.asarray(wf.calculate_discount_rate(
            pure_rate_of_social_time_preference=wf.pure_rate_of_social_time_preference,
            model_time_horizon=lens_model.time_horizon.model_time_horizon,
            timestep=lens_model.time_horizon.timestep,
        ))

        _, _, _, welfare_loss_damage = wf.calculate_welfare(
            physical_damage,
            welfare_loss=True,
        )
        welfare_loss_damage_scalar = float(np.asarray(welfare_loss_damage).squeeze())

        for reg_idx, region_name in region_info:
            damage = np.asarray(physical_damage[reg_idx]).squeeze()
            if damage.ndim > 1:
                damage = damage.mean(axis=-1)
            plotted_value = damage * discount

            for year, damage_value, discount_value, plotted in zip(years_damage, damage, discount, plotted_value):
                records.append({
                    "year": int(year),
                    "region": region_name,
                    "welfare_function": lens_label,
                    "rho": float(wf.pure_rate_of_social_time_preference),
                    "eta": float(wf.elasticity_of_marginal_utility_of_consumption),
                    "inequality_aversion": float(wf.inequality_aversion),
                    "sufficiency_threshold": float(wf.sufficiency_threshold),
                    "egality_strictness": float(wf.egality_strictness),
                    "physical_damage_per_capita": float(damage_value),
                    "discount_factor": float(discount_value),
                    "plotted_value": float(plotted),
                    "plotted_value_description": "physical_damage_per_capita multiplied by JUSTICE welfare_function.calculate_discount_rate; regional time-discounting visibility, not full welfare aggregation",
                    "welfare_loss_damage_scalar": welfare_loss_damage_scalar,
                    "model_native_welfare_loss_note": "scalar from welfare_function.calculate_welfare(physical_damage, welfare_loss=True)",
                })

visibility_df = pd.DataFrame.from_records(records)
visibility_df.to_csv(visibility_csv, index=False)

parameter_summary = (
    visibility_df[[
        "welfare_function", "rho", "eta", "inequality_aversion",
        "sufficiency_threshold", "egality_strictness", "welfare_loss_damage_scalar"
    ]]
    .drop_duplicates()
    .sort_values("welfare_function")
)
print(f"Saved regional damage-visibility data to {visibility_csv}")
display(parameter_summary.round(6))
```

```python
# Plot the saved Prioritarian-vs-Utilitarian damage-visibility data for one focal Actor 4 region.
visibility_csv = RESULTS_ROOT / "final_project" / "prioritarian_vs_utilitarian_regional_damage_visibility.csv"
visibility_df = pd.read_csv(visibility_csv)

lens_order = ["Prioritarian", "Utilitarian"]
lens_colors = {
    "Prioritarian": "#1f77b4",
    "Utilitarian": "#ff7f0e",
}
focal_region = REGION_LABEL_RSAS
region_df = visibility_df[visibility_df["region"] == focal_region].copy()
if region_df.empty:
    raise ValueError(f"No damage-visibility data found for focal region: {focal_region}")

physical = (
    region_df[region_df["welfare_function"] == "Prioritarian"]
    .sort_values("year")
)
years = physical["year"].to_numpy()
physical_damage = physical["physical_damage_per_capita"].to_numpy()

overlap_report = []
fig, ax = plt.subplots(figsize=(8.2, 5.2))

# Filled visible-damage areas first, without legend labels; the legend should explain the curves, not the fills.
for lens in lens_order:
    lens_df = region_df[region_df["welfare_function"] == lens].sort_values("year")
    rho = lens_df["rho"].iloc[0]
    visible_damage = lens_df["plotted_value"].to_numpy()
    color = lens_colors[lens]
    alpha = 0.22 if lens == "Prioritarian" else 0.32
    line_width = 1.8 if lens == "Prioritarian" else 2.3

    ax.fill_between(
        lens_df["year"].to_numpy(),
        np.zeros(len(lens_df)),
        visible_damage,
        color=color,
        alpha=alpha,
        linewidth=0,
        zorder=1 if lens == "Prioritarian" else 2,
    )
    ax.plot(
        lens_df["year"],
        visible_damage,
        color=color,
        lw=line_width,
        alpha=0.95,
        label=f"{lens} visible damage (rho={rho:.3f})",
        zorder=4 if lens == "Prioritarian" else 5,
    )

    if lens == "Prioritarian":
        max_abs_diff = float(np.max(np.abs(visible_damage - physical_damage)))
        max_ref = float(np.max(np.abs(physical_damage)))
        near_identical = bool(np.allclose(visible_damage, physical_damage, rtol=1e-10, atol=1e-12))
        overlap_report.append({
            "region": focal_region,
            "max_abs_difference": max_abs_diff,
            "max_physical_damage": max_ref,
            "near_identical": near_identical,
        })

# Draw physical damage reference last with high zorder so it remains visible even when it overlaps Prioritarian.
ax.plot(
    years,
    physical_damage,
    color="#4f4f4f",
    lw=1.55,
    ls=(0, (4, 3)),
    alpha=0.95,
    label="Physical damage reference",
    zorder=7,
)

ax.set_title(f"Discount-weighted regional climate damage: {focal_region}", fontsize=12)
ax.set_xlabel("Year")
ax.set_ylabel("Damage cost per capita")
ax.grid(alpha=0.22, zorder=0)
handles, labels = ax.get_legend_handles_labels()
order = [2, 0, 1]  # physical reference, Prioritarian, Utilitarian
ax.legend([handles[i] for i in order], [labels[i] for i in order], loc="upper left", frameon=False, fontsize=9)
sns.despine(ax=ax)

fig.text(
    0.5,
    -0.02,
    "Filled areas show visible damage under each discounting assumption. The grey reference is drawn on top; its overlap with the blue curve is substantive, not a plotting error.",
    ha="center",
    fontsize=9,
    color="#444444",
)
plt.tight_layout(rect=[0, 0.04, 1, 1])
img_path = FINAL_PLOTS_DIR / "fp_discount_weighted_damage_prioritarian_vs_utilitarian.png"
plt.savefig(img_path, dpi=150, bbox_inches="tight")
plt.show()
print(f"Saved to {img_path}")
print("Prioritarian vs physical-reference overlap check:")
display(pd.DataFrame(overlap_report))
```

**Interpretation - discount-weighted regional damage.**

## A1/A2 - Uncertainty and sensitivity

Assignments 1 and 2 establish the first premise of the final analysis: uncertainty is not one thing. The project distinguishes:

- **Physical uncertainty**, especially climate sensitivity and socioeconomic/emissions pathway assumptions, which directly govern temperature trajectories and threshold exceedance.
- **Normative uncertainty**, especially `rho`, `eta`, and `delta`, which primarily governs welfare aggregation and intergenerational valuation, and can also interact indirectly with economic feedbacks in the model.

For Actor 4, this distinction matters because physical uncertainty determines how fragile climate safety is, while normative uncertainty determines how seriously the optimiser registers future damages when that safety is breached.

For the Youth Coalition, the important lesson is not only which inputs dominate variance. It is that normative parameters shape welfare visibility and therefore policy advice. A policy can look acceptable under one welfare framing while still leaving future generations exposed under the physical threshold metrics.

## A5/A6 - Optimisation and convergence

Assignments 5 and 6 move from problem formulation to many-objective optimisation and convergence assessment. The workflow uses multiple 50,000-NFE MOEA runs for each welfare lens and combines them into cross-seed reference sets. Multiple seeds matter because a single stochastic MOEA seed is not a reliable representative of the Pareto front in this high-dimensional RBF policy space.

This section therefore treats the reference sets as decision-support evidence, not as mathematically final optima. The aim is to compare the policy landscapes produced by the Prioritarian and Utilitarian framings while remaining honest about convergence limits.


```python
# Assignment 5 synthesis: inventory completed optimisation runs.
from pathlib import Path
import os
import re
import numpy as np
import pandas as pd
from IPython.display import Image, display

# Reuse the portable project root from the setup cell.
if "PROJECT_ROOT" not in globals():
    PROJECT_ROOT = find_project_root()
RESULTS_ROOT = PROJECT_ROOT / "results"
PLOTS_DIR = PROJECT_ROOT / "plots"

OBJECTIVE_COLS = [
    "welfare",
    "fraction_above_threshold",
    "welfare_loss_damage",
    "welfare_loss_abatement",
]

run_rows = []
run_pattern = re.compile(r"^(PRIORITARIAN|UTILITARIAN)_(\d+)_(\d+)$")
for run_dir in sorted(RESULTS_ROOT.iterdir() if RESULTS_ROOT.exists() else []):
    if not run_dir.is_dir():
        continue
    match = run_pattern.match(run_dir.name)
    if not match:
        continue
    welfare_lens, nfe, seed = match.groups()
    pareto_path = run_dir / f"pareto_front_{seed}.csv"
    conv_path = run_dir / f"convergence_{seed}.csv"
    archive_path = run_dir / f"{run_dir.name}.tar.gz"
    n_pareto = len(pd.read_csv(pareto_path)) if pareto_path.exists() else np.nan
    n_conv = len(pd.read_csv(conv_path)) if conv_path.exists() else np.nan
    run_rows.append({
        "welfare_lens": welfare_lens,
        "nfe": int(nfe),
        "seed": int(seed),
        "run_name": run_dir.name,
        "pareto_solutions": n_pareto,
        "convergence_checkpoints": n_conv,
        "archive_present": archive_path.exists(),
    })

run_inventory = pd.DataFrame(run_rows)
if run_inventory.empty:
    raise FileNotFoundError(f"No optimisation run folders found under {RESULTS_ROOT}")

display(
    run_inventory
    .sort_values(["welfare_lens", "seed"])
    .reset_index(drop=True)
)

summary = (
    run_inventory.groupby("welfare_lens")
    .agg(
        runs=("run_name", "count"),
        total_pareto_solutions=("pareto_solutions", "sum"),
        min_solutions=("pareto_solutions", "min"),
        max_solutions=("pareto_solutions", "max"),
    )
    .reset_index()
)
display(summary)
```

## A8 - Robustness

Assignment 8 is the main robustness analysis. It re-evaluates the selected P0-P4 policies across the saved uncertainty design and asks whether each policy is acceptable under stress, not just whether it sits attractively on a nominal Pareto front.

The central evidence is:

- **Satisficing:** how often a policy meets explicit climate and welfare-loss thresholds.
- **Minimax regret:** how bad a policy's Actor 4 score can become relative to the best available selected policy in each scenario.

This section therefore carries the final robustness claim. The ECS diagnostic above is only a supporting bridge and should not be read as the robustness analysis.


```python
# Assignment 8 synthesis: load completed robustness cache.
reeval_path = RESULTS_ROOT / "reeval_actor4_prioritarian_eval_5p_50s.npy"
experiments_path = RESULTS_ROOT / "reeval_actor4_prioritarian_eval_5p_50s_experiments.csv"

if not reeval_path.exists():
    raise FileNotFoundError(f"Missing robustness array: {reeval_path}")
if not experiments_path.exists():
    raise FileNotFoundError(f"Missing robustness experiments CSV: {experiments_path}")

robustness_array = np.load(reeval_path)
robustness_experiments = pd.read_csv(experiments_path)
OUTCOME_COLS = [
    "welfare",
    "years_above_2C",
    "welfare_loss_damage",
    "welfare_loss_abatement",
]

print(f"Robustness array shape: {robustness_array.shape}")
print(f"NaN entries: {np.isnan(robustness_array).sum()}")
print(f"Experiments rows: {len(robustness_experiments)}")

if robustness_array.shape != (5, 50, 4):
    print("Warning: expected robustness array shape (5, 50, 4).")
```

```python
# Robustness outcome summary table.
policy_meta = selected_policy_table[["policy", "selection_label", "welfare_lens"]].copy()
summary_rows = []
for pi, row in policy_meta.iterrows():
    values = robustness_array[pi, :, :]
    entry = {
        "policy": row["policy"],
        "selection_label": row["selection_label"],
        "welfare_lens": row["welfare_lens"],
    }
    for oi, outcome in enumerate(OUTCOME_COLS):
        entry[f"{outcome}_median"] = np.nanmedian(values[:, oi])
        entry[f"{outcome}_p90"] = np.nanpercentile(values[:, oi], 90)
        entry[f"{outcome}_max"] = np.nanmax(values[:, oi])
    summary_rows.append(entry)

robustness_summary = pd.DataFrame(summary_rows)
summary_display_cols = [
    "policy",
    "selection_label",
    "welfare_lens",
    "welfare_median",
    "welfare_p90",
    "years_above_2C_median",
    "years_above_2C_max",
    "welfare_loss_damage_median",
    "welfare_loss_damage_p90",
    "welfare_loss_abatement_median",
    "welfare_loss_abatement_p90",
]

display(robustness_summary[summary_display_cols].round(3))
```

```python
# Satisficing and minimax regret metrics used for Actor 4 interpretation.
# Thresholds follow the completed Assignment 8 analysis.
SATISFICING_THRESHOLDS = {
    "years_above_2C": 0.000,
    "welfare_loss_damage": 96181.812,
    "welfare_loss_abatement": 95312.143,
    "welfare": 98058.994,
}

satisficing_rows = []
for pi, row in policy_meta.iterrows():
    values = robustness_array[pi, :, :]
    df = pd.DataFrame(values, columns=OUTCOME_COLS)
    satisficing_mask = np.ones(len(df), dtype=bool)
    for outcome, threshold in SATISFICING_THRESHOLDS.items():
        satisficing_mask &= df[outcome] <= threshold
    satisficing_rows.append({
        "policy": row["policy"],
        "selection_label": row["selection_label"],
        "welfare_lens": row["welfare_lens"],
        "satisficing_score": satisficing_mask.mean(),
        "n_satisficing": int(satisficing_mask.sum()),
        "n_scenarios": len(df),
        "median_years_above_2C": df["years_above_2C"].median(),
        "max_years_above_2C": df["years_above_2C"].max(),
    })

satisficing_table = pd.DataFrame(satisficing_rows)
display(satisficing_table.round(4))

# Actor 4 regret score: normalise each minimised outcome across all policies/scenarios,
# weight climate safety most strongly, and compare each policy to the best policy per scenario.
weights = {
    "years_above_2C": 0.50,
    "welfare": 0.20,
    "welfare_loss_damage": 0.20,
    "welfare_loss_abatement": 0.10,
}
flat = robustness_array.reshape(-1, robustness_array.shape[-1])
mins = np.nanmin(flat, axis=0)
maxs = np.nanmax(flat, axis=0)
ranges = np.where(maxs - mins == 0, 1, maxs - mins)
normalised = (robustness_array - mins) / ranges

score = np.zeros(robustness_array.shape[:2])
for oi, outcome in enumerate(OUTCOME_COLS):
    score += weights[outcome] * normalised[:, :, oi]

best_per_scenario = np.nanmin(score, axis=0)
regret = score - best_per_scenario

regret_rows = []
for pi, row in policy_meta.iterrows():
    regret_rows.append({
        "policy": row["policy"],
        "selection_label": row["selection_label"],
        "welfare_lens": row["welfare_lens"],
        "max_actor4_regret": np.nanmax(regret[pi, :]),
        "median_actor4_regret": np.nanmedian(regret[pi, :]),
        "p90_actor4_regret": np.nanpercentile(regret[pi, :], 90),
    })

regret_table = pd.DataFrame(regret_rows).sort_values("max_actor4_regret")
display(regret_table.round(4))
```

### Direct answers to the Actor 4 analytical mandate

**1. Pareto shift under low vs high discounting.** The low-discount Prioritarian run and the high-discount Utilitarian run produce different Pareto-front trade-offs. Under the high-discount framing, future damages count for less in the objective calculation, so policies with lower near-term abatement burden or delayed mitigation can look more acceptable. Politically, that benefits actors who prefer lower immediate costs, slower mitigation, or greater reliance on future adaptation, compensation, or technological correction.

**2. Fraction above threshold at 2 degC and the 1.5 degC red line.** The Actor 4 prompt specifically asks for the `fraction_of_ensemble_above_threshold` comparison at 2 degC, so the final notebook retains the 2 degC threshold as the model-comparison and stress-test metric for the saved A5-A8 workflow. Actor 4 does not treat 2 degC as success. The separate 1.5 degC diagnostic overlay shows that all selected policies P0-P4 fail the Youth Coalition's political red line: none has any scenario with zero years above 1.5 degC. This reveals a central negotiation problem: actors can appear to make progress under a 2 degC-oriented metric while still failing intergenerational climate safety.

**3. Combined Pareto-front interpretation.** Showing the Prioritarian and Utilitarian Pareto fronts together makes the normative shift visible. The Prioritarian / near-zero-discount framing keeps future damages and vulnerable regions visible, so stronger mitigation is easier to justify. The Utilitarian / high-discount framing gives more weight to present aggregate welfare and makes delayed or weaker mitigation appear less costly. For Actor 4, the shift between fronts is evidence that the dispute is not only technical; it is also about whose welfare is allowed to disappear from the calculation.

## Political advice

The final recommendation is political as well as technical. Actor 4 should present P0 as the most defensible least-bad modelled compromise among the evaluated options, not as a complete climate solution.

The key advice is:

- **Near-zero discounting is a minimum condition for fair welfare accounting.** It keeps future damages and vulnerable populations visible in the optimisation problem.
- **Near-zero discounting is not sufficient by itself to guarantee climate safety.** Even the Actor-aligned selected policy fails the Youth Coalition's 1.5 degC benchmark in the diagnostic overlay.
- **High-discount modelling should be disclosed and challenged in negotiations.** It can make delayed damages and weaker mitigation appear more acceptable by changing what the objective function values.
- **The feasible policy space remains insufficient for the Youth Coalition mandate.** P0 is defensible only as the least-bad option within the tested A5-A8 policy set.

The negotiation message is therefore: adopt low/zero pure time preference as an ethical baseline, treat 2 degC-oriented performance as insufficient for youth climate safety, and use P0 only as evidence of the strongest available compromise from this model run while demanding a broader and more protective policy search.
